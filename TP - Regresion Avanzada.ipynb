```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

grasacerdos= pd.read_excel("grasacerdos.xlsx")

x=grasacerdos["PV"]
y=grasacerdos["EGD"]

plt.scatter(x, y, marker='o')

# Personalizar el gráfico
plt.title('Diagrama de Dispersión')
plt.xlabel('Peso Vivo')
plt.ylabel('Espesor de grasa dorsal')
plt.grid(False)
plt.legend()

# Mostrar el gráfico
plt.show()  

# Coeficiente de correlacion
r = np.corrcoef(x, y)[0, 1]
print("r:", r)
```

 Ejercicio 1.3. El archivo peso_edad_colest.xlsx disponible en contiene registros correspondientes a 25 individuos respecto de su peso, su edad y el nivel de colesterol total en sangre.

```python
import pandas as pd
import matplotlib.pyplot as plt

df_col = pd.read_excel("peso_edad_colest.xlsx")

print(df_col)

# Configurar dos gráficos lado a lado
fig, axs = plt.subplots(1, 2, figsize=(12, 5))

# Colesterol vs Edad
axs[0].scatter(df_col["edad"], df_col["colest"], color='blue')
axs[0].set_title("Colesterol vs Edad")
axs[0].set_xlabel("Edad")
axs[0].set_ylabel("Colesterol")

# Colesterol vs Peso
axs[1].scatter(df_col["peso"], df_col["colest"], color='green')
axs[1].set_title("Colesterol vs Peso")
axs[1].set_xlabel("Peso")
axs[1].set_ylabel("Colesterol")

plt.tight_layout()
plt.show()
```

Aquí están los diagramas de dispersión:

👉 Conclusión: El modelo lineal parece más adecuado para la relación colesterol vs edad que para colesterol vs peso.

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

# Cargar el archivo Excel
file_path = "peso_edad_colest.xlsx"
df = pd.read_excel(file_path)

# Definir variables
X = df[['edad']]
y = df['colest']

# Ajustar modelo lineal
modelo = LinearRegression()
modelo.fit(X, y)

# Obtener coeficientes
intercepto = modelo.intercept_
pendiente = modelo.coef_[0]

(intercepto, pendiente)
```

Los coeficientes estimados del modelo lineal que predice el colesterol en función de la edad son:

𝑦^=95.50 + 5.67⋅edad

Intercepto (β₀) = 95.50
Pendiente (β₁) = 5.67

Esto significa que por cada año adicional de edad, el colesterol aumenta en promedio aproximadamente 5.67 unidades.

```python
import statsmodels.api as sm

# Agregar constante para el intercepto
X_const = sm.add_constant(X)

# Ajustar modelo con statsmodels para obtener IC y test de Wald
modelo_sm = sm.OLS(y, X_const).fit()

# Resumen del modelo
modelo_sm.summary()
```

Coeficientes estimados con intervalos de confianza al 95%

            Coeficiente	Estimado	IC 95% Inferior	IC 95% Superior	p-valor (Test de Wald)
Intercepto	95.50	                41.19	                 149.81	                 0.001
Edad	    5.67	                4.36	                 6.98	               < 0.001

Análisis

Ambos p-valores son muy bajos, lo que significa que tanto el intercepto como la pendiente son significativamente distintos de cero → Test de Wald para ambos coeficientes rechaza la hipótesis nula.

El intervalo de confianza del 95% para la pendiente no incluye el 0, lo que confirma que hay una relación significativa entre edad y colesterol.

El R² = 0.776, lo que indica que un 77.6% de la variabilidad del colesterol se explica por la edad.

👉 Asociación entre los test: el test de regresión (F = 79.87, p < 0.001) indica que el modelo completo es significativo, lo cual concuerda con los resultados del test de Wald para el coeficiente de edad. Ambos apuntan a que hay una asociación estadísticamente significativa.

```python
# Estimar E(Y) para edades 25 y 48

edad_25 = modelo.predict([[25]])[0]
edad_48 = modelo.predict([[48]])[0]

(edad_25, edad_48)
```

 # 1.3. Transformación de Variables

 Ejercicio 1.4. 
 
 Una empresa desarrolló un sistema de energía solar para calentar el agua para una caldera que es parte del sistema de energía del proceso
 productivo. Existe el interés de controlar la estabilidad del sistema, para ello se monitorea el mismo y se registran los datos cada hora. Los datos se encuentran disponibles en el archivo energia.xlsx

 (a) Realizar el diagrama de dispersión y evaluar si un modelo de regresión
 lineal es adecuado.

 (b) Estimar un modelo lineal y verificar la normalidad de los residuos del
 mismo.

 (c) En caso de rechazar este supuesto buscar una transformación lineal para
 este modelo y aplicarla.

 (d) Realizar el análisis diagnóstico del nuevo modelo y estimar un intervalo de confianza y un intervalo de predicción para 27.5 hs con ambos modelos. Comparar los intervalos
 

```python
import pandas as pd
import matplotlib.pyplot as plt

df=pd.read_excel("energia.xlsx")

print(df)

plt.figure(figsize=(12, 5))
plt.scatter(df["Hora"],df["Energía"])
plt.xlabel("Hora")
plt.ylabel("Energía")
plt.title("Energía vs Hora")

plt.tight_layout()
plt.show()
```

Visualmente, se observa una tendencia descendente, lo que sugiere que podría haber una relación lineal negativa entre ambas variables.

```python
import statsmodels.api as sm
import seaborn as sns

X_const=sm.add_constant(df["Hora"])
y=df["Energía"]

modelo_sm=sm.OLS(y,X_const).fit()

modelo_summary= modelo_sm.summary()

residuos=modelo_sm.resid

#Verifico Normalidad

plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
sns.histplot(residuos,kde=True)
plt.title("Distribución de residuos")

plt.subplot(1,2,2)
stats.probplot(residuos, dist="norm", plot=plt)
plt.title("QQ-Plot de los residuos")

plt.tight_layout()
plt.show()

# Test de normalidad de Shapiro-Wilk
shapiro_test = stats.shapiro(residuos)

print(modelo_summary)
print(shapiro_test)
print(shapiro_test.pvalue)
```

✅ Resultados del punto (b)
📌 Modelo lineal ajustado:
Ecuación estimada:

Energia=491.49+0.52⋅Hora

Coeficiente de determinación R²: 0.006 → el modelo explica solo el 0.6% de la variabilidad.

p-valor del coeficiente de Hora: 0.597 → no es significativo.

📊 Normalidad de los residuos:
Test de Shapiro-Wilk:

Estadístico = 0.93

p-valor = 0.0067 → Rechazamos la hipótesis nula de normalidad al 5% de significación.

Los gráficos también muestran una asimetría a la derecha.

```python
import numpy as np

# Crear la variable transformada (logarítmica)
df['log_Energía'] = np.log(df['Energía'])

# Definir variables para el nuevo modelo
X_log = sm.add_constant(df['Hora'])
y_log = df['log_Energía']

# Ajustar nuevo modelo
modelo_log = sm.OLS(y_log, X_log).fit()

# Residuos del modelo logarítmico
residuos_log = modelo_log.resid

# Histograma de residuos logarítmicos
plt.figure(figsize=(8, 4))
sns.histplot(residuos_log, kde=True, color='mediumseagreen')
plt.title('Histograma de Residuos (Modelo Logarítmico)')
plt.xlabel('Residuos')
plt.grid(True)
plt.show()

# QQ plot
sm.qqplot(residuos_log, line='s')
plt.title('QQ Plot de Residuos (Modelo Logarítmico)')
plt.grid(True)
plt.show()

# Test de Shapiro-Wilk para modelo transformado
shapiro_log = stats.shapiro(residuos_log)

modelo_log.summary(), shapiro_log.statistic, shapiro_log.pvalue
```

✅ Resultados del punto (c): Modelo con Transformación Logarítmica
📌 Modelo logarítmico ajustado:
Ecuación estimada:

log(Energia)=6.194+0.0006⋅Hora

R² = 0.002 → aún muy bajo, el modelo explica apenas el 0.2% de la variabilidad.

p-valor del coeficiente de Hora = 0.761 → no significativo.

📊 Normalidad de los residuos:
Test de Shapiro-Wilk:

Estadístico 
𝑊=0.964
p-valor = 0.145 → no se rechaza la hipótesis de normalidad ✅

🧠 Interpretación:
La transformación logarítmica mejora la normalidad de los residuos. Aunque el modelo no mejora en cuanto a capacidad explicativa, sí cumple mejor los supuestos del modelo lineal clásico.



```python
# Modelos original y logarítmico
X = sm.add_constant(df['Hora'])
y = df['Energía']
modelo = sm.OLS(y, X).fit()

df['log_Energía'] = np.log(df['Energía'])
y_log = df['log_Energía']
modelo_log = sm.OLS(y_log, X).fit()

# Predicción para 27.5 horas en ambos modelos
hora_275 = pd.DataFrame({'const': [1], 'Hora': [27.5]})

# Intervalos modelo original
pred_orig = modelo.get_prediction(hora_275).summary_frame(alpha=0.05)

# Intervalos modelo logarítmico
pred_log = modelo_log.get_prediction(hora_275).summary_frame(alpha=0.05)

# Transformar de nuevo a escala original los intervalos logarítmicos
pred_log[['mean', 'mean_ci_lower', 'mean_ci_upper', 'obs_ci_lower', 'obs_ci_upper']] = \
    np.exp(pred_log[['mean', 'mean_ci_lower', 'mean_ci_upper', 'obs_ci_lower', 'obs_ci_upper']])

# Combinar y mostrar resultados
pred_orig['Modelo'] = 'Original'
pred_log['Modelo'] = 'Logarítmico'

intervalos = pd.concat([pred_orig, pred_log])
print(intervalos[['Modelo', 'mean', 'mean_ci_lower', 'mean_ci_upper', 'obs_ci_lower', 'obs_ci_upper']])
```

📌 Intervalo de Confianza → sobre la media esperada
mean_ci_lower y mean_ci_upper
→ Límites inferior y superior del intervalo de confianza al 95% para el valor esperado de la variable dependiente (la media de Y para un valor dado de X).

📌 Intervalo de Predicción → sobre una nueva observación
obs_ci_lower y obs_ci_upper
→ Límites del intervalo de predicción al 95%, es decir, el rango donde esperamos que caiga una nueva observación individual de Y para ese valor de X.

🧠 Diferencia clave:
El intervalo de confianza es más angosto porque solo tiene en cuenta la incertidumbre del estimador de la media.

El intervalo de predicción es más ancho porque incluye, además, la variabilidad natural del fenómeno (la dispersión de los residuos).

```python
def graficoMatrizCorrelacion(corr_matrix):
    # Matriz de correlacion con los valores

    import numpy as np

    # Crear el gráfico
    fig, ax = plt.subplots(figsize=(8, 6))
    cax = ax.matshow(corr_matrix, cmap='coolwarm')

    # Agregar la barra de color
    fig.colorbar(cax)

    # Etiquetas
    ax.set_xticks(np.arange(len(corr_matrix.columns)))
    ax.set_yticks(np.arange(len(corr_matrix.columns)))
    ax.set_xticklabels(corr_matrix.columns, rotation=45, ha='left')
    ax.set_yticklabels(corr_matrix.columns)

    # Mostrar los valores de correlación
    for i in range(len(corr_matrix.columns)):
        for j in range(len(corr_matrix.columns)):
            value = corr_matrix.iloc[i, j]
            ax.text(j, i, f"{value:.2f}", va='center', ha='center', color='black', fontsize=9)

    plt.title('Matriz de Correlación', pad=20)
    plt.tight_layout()
    plt.show()
```

```python
def graficoResiduosNormalidad(modelo_sm):
    
    import statsmodels.api as sm
    import scipy.stats as stats
    from statsmodels.graphics.regressionplots import influence_plot
    import seaborn as sns
    
    residuos=modelo_sm.resid
    
    plt.figure(figsize=(12,5))

    plt.subplot(1,2,1)
    sns.histplot(residuos,kde=True)
    plt.title("Distribución de residuos")

    plt.subplot(1,2,2)
    stats.probplot(residuos, dist="norm", plot=plt)
    plt.title("QQ-Plot de los residuos")
    
    plt.tight_layout()
    plt.show() 
    
    # Residuos vs leverage
    
    # Crear el gráfico sin etiquetas de puntos
    fig, ax = plt.subplots(figsize=(12, 5))
    influence_plot(modelo_sm, criterion="cooks", ax=ax, alpha=0.5)

    # Ocultar etiquetas de puntos (modificando texto)
    for text in ax.texts:
        text.set_visible(False)

    plt.title("Residuos estandarizados vs leverage")
    plt.tight_layout()
    plt.show()
    
    # Test de normalidad de Shapiro-Wilk
    shapiro_test = stats.shapiro(residuos)

    print(shapiro_test)
    print(shapiro_test.pvalue)
```

```python
import statsmodels.api as sm
import scipy.stats as stats

# Definir variables
X = df["distancia"]
y = df["precio"]

# Agregar constante para el intercepto
X_const = sm.add_constant(X)

# Ajustar modelo con statsmodels para obtener IC y test de Wald
modelo_sm = sm.OLS(y, X_const).fit()

# Resumen del modelo
modelo_sm.summary()

#Verifico Normalidad

graficoResiduosNormalidad(modelo_sm)
```

Valores altos de leverage indican que una observación tiene un valor extremo en alguna variable independiente.

No significa necesariamente que sea un error o un valor mal estimado, pero sí que puede influir fuertemente en la línea de regresión.

🔎 Una observación con alto leverage está lejos del centro de los datos (en X), no necesariamente en Y.

🧮 ¿Cómo se interpreta el gráfico Residuos estandarizados vs Leverage?
Este gráfico muestra:

En el eje X: el leverage de cada observación (ℎ𝑖)

En el eje Y: el residuo studentizado o estandarizado (una versión normalizada del error)

Se agregan curvas de distancia de Cook, que indican qué puntos influyen mucho en el modelo completo.

📊 ¿Qué se busca identificar?
🔸 Puntos con alto leverage (lejanos en X)
Tienen valores extremos en las variables predictoras. Están hacia la derecha del gráfico.

🔸 Puntos con residuos grandes (lejanos en Y)
Tienen errores altos en la predicción. Están lejos del eje horizontal.

🔸 Puntos influyentes
Tienen ambas cosas: están lejos en X y tienen errores grandes → pueden cambiar la forma de la recta de regresión si los quitamos.

Estos son los que suelen estar fuera de las líneas de Cook's distance.

🧠 ¿Qué es la Distancia de Cook?
La distancia de Cook mide cuánto cambiaría el modelo si elimináramos una observación.

Se representa con curvas en el gráfico.

Si un punto está fuera de la curva de Cook 0.5 o 1, suele considerarse influyente.

✅ ¿Qué hacer con los puntos influyentes?
Revisarlos: ver si hay errores de carga, errores tipográficos o casos especiales.

Decidir: si se justifican, los dejás. Si son errores o casos no representativos, podés eliminarlos o analizar por separado.

Opcional: comparar el modelo con y sin esos puntos para ver su impacto.



```python
import statsmodels.api as sm
from statsmodels.stats.stattools import durbin_watson
from statsmodels.stats.diagnostic import het_breuschpagan

# Definir X y y (modelo simple: precio vs distancia)
X = sm.add_constant(df['distancia'])
y = df['precio']
modelo = sm.OLS(y, X).fit()

# Test de Durbin-Watson
dw = durbin_watson(modelo.resid)
print(f"Durbin-Watson: {dw:.3f}")

# Test de Breusch-Pagan
bp_test = het_breuschpagan(modelo.resid, X)
bp_labels = ['Estadístico LM', 'p-valor LM', 'Estadístico F', 'p-valor F']
print("\nBreusch-Pagan test:")
for label, value in zip(bp_labels, bp_test):
    print(f"{label}: {value:.4f}")
```

Hipótesis alternativa: la varianza cambia (heterocedasticidad).

Si el p-valor es bajo (< 0.05), se rechaza la homocedasticidad.

```python
# Cargar el archivo para realizar el análisis completo
import pandas as pd
import statsmodels.api as sm
import numpy as np
from scipy.stats import f 

# Obtener objeto de influencia para métricas de diagnóstico
influence = modelo.get_influence()

# 1. RESIDUOS ESTUDENTIZADOS INTERNOS
# Se utilizan para detectar outliers: observaciones cuyo error de predicción es inusualmente alto o bajo
res_stu = influence.resid_studentized_internal
outliers_residuales = np.where(np.abs(res_stu) > 3)[0]

# 2. DFBETAS
# Mide cuánto cambiaría un coeficiente del modelo si eliminamos una observación
# Se analiza el segundo coeficiente (pendiente de 'distancia'), que está en la columna 1
dfbetas = influence.dfbetas
dfbetas_beta1_grandes = np.where(np.abs(dfbetas[:, 1]) > 1)[0]  # criterio usual: > 1

# 3. DFFITS
# Mide cuánto cambia el valor ajustado (predicción) si quitamos una observación
dffits = influence.dffits[0]
n = len(df)
p = X.shape[1]  # número de parámetros (constante + 1 predictor)
corte_dffits = 2 * np.sqrt(p / n)
dffits_grandes = np.where(np.abs(dffits) > corte_dffits)[0]

# 4. LEVERAGE
# Mide cuán "extrema" es una observación en las X (cuánto apalanca el modelo)
leverage = influence.hat_matrix_diag
lev_mayor_0_2 = np.where(leverage > 0.2)[0]
lev_mayor_2pn = np.where(leverage > 2 * p / n)[0]

# 5. DISTANCIA DE COOK
# Evalúa la influencia total de una observación: mezcla leverage + residuo
cooks_d = influence.cooks_distance[0]
corte_cooks = 4 / n
cooks_grandes_4n = np.where(cooks_d > corte_cooks)[0]
corte_cooks_f = f.ppf(0.5, p - 1, n - p)
cooks_grandes_fcrit = np.where(cooks_d > corte_cooks_f)[0]

# Identificar observaciones que cumplen múltiples criterios
conjunto_outliers = set(outliers_residuales)
conjunto_cooks = set(cooks_grandes_4n)
conjunto_dffits = set(dffits_grandes)
conjunto_leverage = set(lev_mayor_2pn)

# Coincidencias entre outliers y puntos influyentes
coinciden_outlier_cooks = conjunto_outliers & conjunto_cooks
coinciden_outlier_dffits = conjunto_outliers & conjunto_dffits
coinciden_cooks_dffits = conjunto_cooks & conjunto_dffits
coinciden_todos = conjunto_outliers & conjunto_cooks & conjunto_dffits

# Consolidar resultados
resumen = {
    "outliers_residuales_(|r|>3)": outliers_residuales.tolist(),
    "dfbetas_beta1_>|1|": dfbetas_beta1_grandes.tolist(),
    "dffits_>_2sqrt(p/n)": dffits_grandes.tolist(),
    "leverage_>_0.2": lev_mayor_0_2.tolist(),
    "leverage_>_2p/n": lev_mayor_2pn.tolist(),
    "cooks_>_4/n": cooks_grandes_4n.tolist(),
    "cooks_>_F(0.5;2,n-2)": cooks_grandes_fcrit.tolist()
}

print(resumen)

# Resultados de análisis de coincidencias
print(f"Coinciden como outlier + Cook: {sorted(coinciden_outlier_cooks)}")
print(f"Coinciden como outlier + DFFITS: {sorted(coinciden_outlier_dffits)}")
print(f"Coinciden como Cook + DFFITS: {sorted(coinciden_cooks_dffits)}")
print(f"Coinciden en los tres: {sorted(coinciden_todos)}")
```

```python
# Estimación de la varianza de los residuos agrupando por valores similares de horas_estudio
# Paso 1: Agrupar por valores únicos de horas de estudio
grupo_varianzas = df.groupby("horas_estudio")["puntaje"].var()

# Paso 2: Mapear a cada fila su varianza estimada
df["varianza_estimada"] = df["horas_estudio"].map(grupo_varianzas)

# Paso 3: Los pesos son inversos a la varianza (mayor varianza = menor peso)
df["peso"] = 1 / df["varianza_estimada"]

# Eliminar observaciones con varianza no definida
df_valid = df.dropna(subset=["varianza_estimada"]).copy()

# Recalcular pesos (inverso de la varianza estimada)
df_valid["peso"] = 1 / df_valid["varianza_estimada"]

# Ajustar el modelo WLS
X_wls = sm.add_constant(df_valid["horas_estudio"])
modelo_wls = sm.WLS(df_valid["puntaje"], X_wls, weights=df_valid["peso"]).fit()

# Mostrar el resumen del modelo WLS
modelo_wls.summary()
```

📌 Interpretación:

* El modelo WLS logra ajustar mucho mejor los datos: el 𝑅2 pasa de 0.52 a 0.92.

* El efecto estimado de las horas de estudio es más fuerte: 7.998 puntos adicionales por cada hora extra.

* La reducción de AIC indica un mejor balance entre calidad de ajuste y complejidad del modelo.
